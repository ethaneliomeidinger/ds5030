In general, you can only bet on 1, 2, 3, 4, 6 slots, 12 slots, or 18 slots. If your bet occurs when the wheel is spun, you gain 36/K-1 where $K$ is the number of slots you bet on; if not, you lose a dollar and get -1.

```python
import random


reds = {1,3,5,7,9,12,14,16,18,19,21,23,25,27,30,32,34,36}
blacks = set(range(1,37)) - reds
greens = {0}

all_numbers = list(range(37))  

def spin():
    """Return one outcome number from 0..36 (uniform)."""
    return random.choice(all_numbers)
```

```python
"""
Prob Space
* Sample Space: (0,1,2,...36) 37 outcomes, each equally likely
* Event Space: 2^37 wow
* Each number is equally likely with prob 1/37. so a roll succeeding has the 
prob K/37 where K is the number of outcomes in the bet set.
"""
```

```python
def result_for_bet(bet_set, outcome):
    """
    bet_set: a set of numbers in [0..36] that the bet covers (size K in {1,2,3,4,6,12,18})
    outcome: the number spun
    Returns the net result for the player: 36/K - 1 if win, else -1.
    """
    K = len(bet_set)
    if outcome in bet_set:
        return 36.0 / K - 1.0
    else:
        return -1.0
```

```python
def simulate_avg(bet_set, trials=1000):
    total = 0.0
    for _ in range(trials):
        outcome = spin()
        total += result_for_bet(bet_set, outcome)
    return total / trials
```

```python
bet_red = reds                                  
bet_odd = {n for n in range(1,37) if n % 2 == 1}  
bet_single_7 = {7}                              


bet_split_1_2 = {1, 2}                         
bet_line_1_to_6 = {1,2,3,4,5,6}                

N = 1000
avg_red = simulate_avg(bet_red, N)
avg_odd = simulate_avg(bet_odd, N)
avg_single7 = simulate_avg(bet_single_7, N)
avg_split = simulate_avg(bet_split_1_2, N)
avg_line = simulate_avg(bet_line_1_to_6, N)

print(f"Average payoff over {N} spins:")
print(f"  Red        (K=18): {avg_red:.4f}")
print(f"  Odd        (K=18): {avg_odd:.4f}")
print(f"  Single 7   (K=1) : {avg_single7:.4f}")
print(f"  Split 1-2  (K=2) : {avg_split:.4f}")
print(f"  Line 1–6   (K=6) : {avg_line:.4f}")
```

### 5. CDF and PDF Basics

- Verify the following functions are distribution functions and compute their density functions. Plot the distribution and density.

1. $F(x) = \begin{cases}
0, & x \le 0 \\
\sqrt{x}, & 0 \le x \le 1 \\
1, & x \ge 1 
\end{cases}
$
2. $F(x) = \dfrac{1}{1+e^{-x}}$
3. For $ a < b < c$,
$
F(x) = \begin{cases}
0, & x \le 0 \\
\frac{(x-a)^2}{(b-a)(b-c)}, & a \le x \le c \\
1 - \frac{(b-x)^2}{(b-a)(b-c)}, & c < x < b \\
1, & x \ge b
\end{cases}
$
4. $ F(x) = \begin{cases}
0, & x <0 \\
1 - e^{-x}, & x>0
\end{cases}
$
5. $ F(x) = \begin{cases}
0, & x \le 0 \\
x, & 0 < x < 1\\
1, & x \ge 1
\end{cases}
$

```python

a, c, b = 0.0, 0.3, 1.0

#cdf
def F1(x):
    x = np.asarray(x)
    out = np.zeros_like(x, dtype=float)
    mask1 = (x >= 0) & (x <= 1)
    out[mask1] = np.sqrt(x[mask1])
    out[x > 1] = 1.0
    return out

def F2(x):
    x = np.asarray(x, dtype=float)
    return 1.0 / (1.0 + np.exp(-x))

def F3(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    m1 = (x >= a) & (x <= c)
    m2 = (x >= c) & (x <= b)
    out[m1] = ((x[m1] - a)**2) / ((b - a) * (c - a))
    out[m2] = 1.0 - ((b - x[m2])**2) / ((b - a) * (b - c))
    out[x >= b] = 1.0
    return out

def F4(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    m = (x > 0)
    out[m] = 1.0 - np.exp(-x[m])
    return out 

def F5(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    m = (x > 0) & (x < 1)
    out[m] = x[m]
    out[x >= 1] = 1.0
    return out

#pdf
def f1(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    m = (x > 0) & (x < 1)
    out[m] = 1.0 / (2.0 * np.sqrt(x[m]))
    return out

def f2(x):
    x = np.asarray(x, dtype=float)
    Fx = F2(x)
    return Fx * (1.0 - Fx)

def f3(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    out[(x > a) & (x < c)] = 2.0 * (x[(x > a) & (x < c)] - a) / ((b - a) * (c - a))
    out[(x > c) & (x < b)] = 2.0 * (b - x[(x > c) & (x < b)]) / ((b - a) * (b - c))
    return out

def f4(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    m = (x > 0)
    out[m] = np.exp(-x[m])
    return out

def f5(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    out[(x > 0) & (x < 1)] = 1.0
    return out

items = [
    ("Item 1", F1, f1, (-0.5, 1.5)),
    ("Item 2", F2, f2, (-6, 6)),
    ("Item 3", F3, f3, (a - 0.2, b + 0.2)),
    ("Item 4", F4, f4, (-0.5, 5.0)),
    ("Item 5", F5, f5, (-0.5, 1.5)),
]

for title, F, f, (xmin, xmax) in items:
    xs = np.linspace(xmin, xmax, 1200)
    plt.figure(figsize=(6,4))
    plt.title(f"{title} — CDF")
    plt.plot(xs, F(xs))
    plt.xlabel("x"); plt.ylabel("F(x)"); plt.ylim(-0.05, 1.05)

    plt.figure(figsize=(6,4))
    plt.title(f"{title} — PDF")
    plt.plot(xs, f(xs))
    plt.xlabel("x"); plt.ylabel("f(x)")


```

You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.


# Supports
* Logistic : all real number
* Exponential : [0 ... inf]
* Neg Binom: 0 and all positive integers
* Categorical: k where p_k > 0

```python
def ecdf(x):
    x_sorted = np.sort(x)
    y = np.arange(1, len(x_sorted)+1) / len(x_sorted)
    return x_sorted, y
```

```python
r, p = 5, 0.4
n = 1000

x = nbinom.rvs(n=r, p=p, size=n, random_state=rng)

k_emp = np.arange(x.max() + 1)
pmf_emp = np.bincount(x, minlength=k_emp.size) / n
pmf_th = nbinom.pmf(k_emp, n=r, p=p)

plt.figure(figsize=(6,4))
plt.stem(k_emp, pmf_th, linefmt='-', markerfmt='o', basefmt=' ', label='theoretical')
plt.plot(k_emp, pmf_emp, 'x', label='empirical (n=1000)')
plt.title(f"Negative Binomial PMF (r={r}, p={p})")
plt.xlabel("k = # failures before r successes")
plt.ylabel("P(X=k)")
plt.legend()

x_sorted = np.sort(x)
ecdf = np.arange(1, n+1) / n
k_cdf = np.arange(0, x.max()+1)
cdf_th = nbinom.cdf(k_cdf, n=r, p=p)

plt.figure(figsize=(6,4))
plt.step(x_sorted, ecdf, where='post', label='ECDF (n=1000)')
plt.step(k_cdf, cdf_th, where='post', label='theoretical CDF')
plt.title("Negative Binomial: ECDF vs Theoretical CDF")
plt.xlabel("k")
plt.ylabel("F(k)")
plt.legend()
```

```python
p_vec = np.array([0.1, 0.3, 0.4, 0.2])   # must sum to 1
K = len(p_vec)
n = 1000

cats = rng.choice(np.arange(1, K+1), size=n, p=p_vec)

# Empirical probs
counts = np.bincount(cats, minlength=K+1)[1:]  # drop index 0
prob_emp = counts / n

plt.figure(figsize=(6,4))
xpos = np.arange(1, K+1)
barw = 0.35
plt.bar(xpos - barw/2, p_vec, width=barw, label='theoretical')
plt.bar(xpos + barw/2, prob_emp, width=barw, label='empirical (n=1000)')
plt.title("Categorical PMF (theoretical vs empirical)")
plt.xlabel("category")
plt.ylabel("probability")
plt.xticks(xpos)
plt.legend()

print(f"Categorical: emp={prob_emp.round(3)}, th={p_vec.round(3)}")
```
